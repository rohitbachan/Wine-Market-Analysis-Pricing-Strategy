```python
# Fill missing price values with median
df['price'] = df['price'].fillna(df['price'].median())

# Check
print("Missing price:", df['price'].isnull().sum())
print("Median price used:", df['price'].median())
```

```python
# Fill missing categorical values
df['country'] = df['country'].fillna('Unknown')
df['variety'] = df['variety'].fillna('Unknown')
```

```python
# Then check:
print(df[['country', 'variety']].isnull().sum())
```

```python
# Fill missing categorical values
categorical_cols = ['designation','province','region_1','region_2','taster_name','taster_twitter_handle']
for col in categorical_cols:
    df[col] = df[col].fillna('Unknown')
```

```python
# Check duplicate rows
duplicate_count = df.duplicated().sum()
print("Duplicate rows:", duplicate_count)
```

```python
#Find unique values in the 'country' column
df['country'].unique()
```

```python
#Find unique values in the 'variety' column
print(df['variety'].unique())
```

```python
#Find unique values in the 'province' column
df['province'].unique()[:50]
```

```python
# Remove leading and trailing spaces from all text columns
text_cols = df.select_dtypes(include='str').columns
for col in text_cols:
    df[col] = df[col].str.strip()
```

```python
# Check the first few rows of the DataFrame to verify that the changes have been applied
print(df[['country', 'province', 'variety']].head())
```

```python
# Create a new column 'price_segment' based on the price ranges
def price_segment(price):
    if price <= 30:
        return 'Low'
    elif price <= 50:
        return 'Medium'
    else:
        return 'Premium'

df['price_segment'] = df['price'].apply(price_segment)
```

```python
# Check the distribution of price segments
print(df['price_segment'].value_counts())
```

```python
# Check the first few rows of the DataFrame to verify that the changes have been applied
print(df[['price', 'price_segment']].head(10))
```

```python
# Create a new column 'value_score' based on the ratio of points to price
df['value_score'] = df['points'] / df['price']
```

```python
# Check the first few rows of the DataFrame to verify that the changes have been applied
print(df[['points', 'price', 'value_score']].head(10))
```

```python
# Check for any remaining null values in the new columns
print(df[['price_segment', 'value_score']].isnull().sum())
```

```python
# Check the distribution of value scores
print(df['value_score'].describe())
```

```python
# Save the cleaned dataset to a new CSV file
df.to_csv('cleaned_wine_dataset.csv',index=False)
```
